# **Project Description**
This project focuses on building a Demand Forecasting system for retail stores using machine learning techniques. The main objective is to predict daily sales for each store based on historical data and various influencing factors such as promotions, holidays, competition, and customer behavior.

The project simulates a real-world business problem where accurate sales predictions can help companies optimize inventory, improve marketing strategies, and make better operational decisions. It includes a complete pipeline starting from data exploration and preprocessing to model training, evaluation, and deployment.

---

# **Data Description**
The dataset contains daily sales records for Rossmann stores in Germany. Each row represents a store’s performance on a specific day, including features such as store ID, date, sales (target), customers, promotions, holidays, store type, and competition-related information.

---

# **Column-by-column description**
1.**Store**
* Unique identifier for each store branch.

2.**DayOfWeek**
* Represents the day of the week (from 1 to 7) on which the sales were recorded, and is used to analyze weekly sales patterns.

3.**Date**
* The specific date of the recorded sales.

4.**Sales**
* Total daily revenue (turnover) generated by the store (Target variable).

5.**Customers**
* Number of customers who made purchases in the store on that day.

6.**Open**
* Indicates whether the store was open (1) or closed (0) on that day.

7.**StateHoliday**
* Indicates if the day was a public/state holiday
* (a = public holiday, b = Easter, c = Christmas, 0 = none).

8.**SchoolHoliday**
* Indicates whether public schools were closed on that day.

9.**StoreType**
* Categorizes the store into one of four different store models (a, b, c, d).
* (A = Hypermarket & B = Small Store & C = Specialty & D = Premium)

10.**Assortment**
* Describes the product assortment level
* (a = basic, b = extra, c = extended).

11.**CompetitionDistance**
* Distance (in meters) to the nearest competitor store.

12.**CompetitionOpenSinceMonth**
* The month when the nearest competitor store opened.

13.**CompetitionOpenSinceYear**
* The year when the nearest competitor store opened.

14.**Promo**
* Indicates whether the store was running a promotion on that day (1 = yes, 0 = no).

15.**Promo2**
* Indicates whether the store is participating in a continuous promotion program (1 = yes, 0 = no).

16.**Promo2SinceYear**
* The year when the store started participating in Promo2.

17.**Promo2SinceWeek**
* The calendar week when the store started participating in Promo2.

18.**PromoInterval**
* Describes the months when Promo2 is activated each year

---

**2. Handling Missing Values**


```python
# Check missing values
df.isnull().sum()
```

```python
# fill missing value in CompetitionDistance with median
df["CompetitionDistance"] = df['CompetitionDistance'].fillna(df['CompetitionDistance'].median())
```

```python
# fill missing values in CompetitionOpenSinceMonth and CompetitionOpenSinceYear with 0
# Notes => 0 represent to unknown value
df["CompetitionOpenSinceMonth"] = df["CompetitionOpenSinceMonth"].fillna(0)
df["CompetitionOpenSinceYear"] = df["CompetitionOpenSinceYear"].fillna(0)
```

```python
# fill missing values with 0 for num_cols and None for cat_cols
df["Promo2SinceWeek"] = df["Promo2SinceWeek"].fillna(0)
df["Promo2SinceYear"] = df["Promo2SinceYear"].fillna(0)
df["PromoInterval"] = df["PromoInterval"].fillna("None")
```

**3. Feature Enginering**

```python
df.head()
```

```python
# Extract some usful features from column Data

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["DayOfMonth"] = df["Date"].dt.day

# column to check if is start and end of the month
df["startmonth"] = df['DayOfMonth'].isin([1,2,3,4,5,6,7]).astype(int)
df["endmonth"] = df['DayOfMonth'].isin([25,26,27,28,29,30,31]).astype(int)

# column to check if Is Weekend
df["IsWeekend"] = df["DayOfWeek"].isin([5,6]).astype(int)
```

```python
# create column to check if CompetitionDate is Known or not
df["CompetitionDateKnown"] = (df["CompetitionOpenSinceYear"] != 0).astype(int)
```

```python
# Detrmine the competition duration (if the data is known)
# Note => errors if found any NaN or 0 pandas convert it to Nat which mean Not a Time
df["CompetitionOpenDate"] = pd.to_datetime(
    dict(
        year = df["CompetitionOpenSinceYear"],
        month = df["CompetitionOpenSinceMonth"],
        day = 1
    ),
    errors = "coerce"
    )
```

```python
# create column detrmine the competition duration and convert it to months
df["CompetitionDurationMonths"] = (df['Date'] - df['CompetitionOpenDate']).dt.days/30
```

```python
# handle missing in CompetitionDurationMonths
df["CompetitionDurationMonths"] = df["CompetitionDurationMonths"].fillna(0)

# Notes => Sometimes CompetitionOpenDate > Date ,so we found negative values
# filter to convert any negative value to 0
df["CompetitionDurationMonths"] = df["CompetitionDurationMonths"].clip(lower=0)
```

```python
df["IsPromo"] = df["Promo"]
df["IsPromo2"] = df["Promo2"]
```

```python
# Drop unimportant columns
df.drop(["Store","Promo", "Promo2","CompetitionOpenDate","Date" ,'Promo2SinceWeek',
       'Promo2SinceYear', 'PromoInterval',"DayOfMonth" ,"CompetitionOpenSinceMonth" ,"CompetitionOpenSinceYear"], axis=1, inplace=True)
```

```python
# filter the rows that open = 1
df = df[df["Open"] == 1]
```

```python
# drop Open
df.drop("Open", axis=1, inplace=True)
```

**1- R² (Coefficient of Determination)** => Score measures how well the model explains the variability in the target variable.
A higher R² value indicates that the model fits the data better and explains a larger proportion of the variance.

**2- MAE (Mean Absolute Error)** => represents the average absolute difference between the actual and predicted values.
It provides a clear interpretation of the average prediction error.

**3- RMSE (Root Mean Squared Error)** => measures the square root of the average squared differences between predictions and actual values.
It penalizes larger errors more heavily, making it sensitive to significant prediction mistakes.

```python
# Import
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
# Evalution on LinearRegression

y_hat_LR = LR.predict(x_test)

r2_LR = r2_score(y_test, y_hat_LR)
rmse_LR = np.sqrt(mean_squared_error(y_test, y_hat_LR))
mae_LR = mean_absolute_error(y_test, y_hat_LR)
```

```python
# Evalution on DecisionTreeRegressor

y_hat_DTR = DTR.predict(x_test)

r2_DTR = r2_score(y_test, y_hat_DTR)
rmse_DTR = np.sqrt(mean_squared_error(y_test, y_hat_DTR))
mae_DTR = mean_absolute_error(y_test, y_hat_DTR)
```

```python

# Evalution on RandomForestRegressor

y_hat_RFR = RFR.predict(x_test)

r2_RFR = r2_score(y_test, y_hat_RFR)
rmse_RFR = np.sqrt(mean_squared_error(y_test, y_hat_RFR))
mae_RFR = mean_absolute_error(y_test, y_hat_RFR)
```

```python
# Evalution on GradientBoostingRegressor

y_hat_GBR = GBR.predict(x_test)

r2_GBR = r2_score(y_test, y_hat_GBR)
rmse_GBR = np.sqrt(mean_squared_error(y_test, y_hat_GBR))
mae_GBR = mean_absolute_error(y_test, y_hat_GBR)
```

```python
# Evalution on XGBRegressor

y_hat_XGB = XGBR.predict(x_test)

r2_XGB = r2_score(y_test, y_hat_XGB)
rmse_XGB = np.sqrt(mean_squared_error(y_test, y_hat_XGB))
mae_XGB = mean_absolute_error(y_test, y_hat_XGB)
```

```python
# Compare Between five  models

results = pd.DataFrame({
    "Model": ["LinearRegression","DecisionTree","RandomForest", "GradientBoosting", "XGBoost"],
    "R2": [r2_LR ,r2_DTR ,r2_RFR, r2_GBR, r2_XGB],
    "RMSE": [rmse_LR ,rmse_DTR ,rmse_RFR, rmse_GBR, rmse_XGB],
    "MAE": [mae_LR ,mae_DTR ,mae_RFR, mae_GBR, mae_XGB]
})

print(results)
```

**4- Final Evalution**

```python
# Code to display and compare between three models

models = {
    "RandomForest": RFR2,
    "GradientBoosting": GBoost2,
    "XGBoost": XGBoost2
}

results = []


for name, model in models.items():

    y_pred = model.predict(x_test)

    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)

    results.append({
        "Model": name,
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae
    })

results = pd.DataFrame(results)

print(results)
```

Although XGBoost achieved a slightly lower MAE, the difference between the models was minimal. Therefore, GradientBoostingRegressor was selected as the final model for deployment and integration into the Streamlit application.

```python
# 1- Divide the features which enter to the model to categorical_features and numerical_features
categorical_features = [
"DayOfWeek",
"SchoolHoliday",
"IsPromo",
"IsPromo2"
]
numerical_features = [
"Customers",
"Year",
"Month",
"CompetitionDurationMonths_log"
]
```

```python
# Import
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
# Create Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", StandardScaler(), numerical_features)
    ]
)
# Create pipline
pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", GBoost2)
])
# train pipline
pipeline.fit(x_train, y_train)
```

```python
# Code to check that pipline train good and compare the result before and after pipline

# test
pred = pipeline.predict(x_test)

mae = mean_absolute_error(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))
r2 = r2_score(y_test, pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib
import numpy as np

# Load model
model = joblib.load("demand_forecasting_model.pkl")

# Page config
st.set_page_config(page_title="Store Sales Prediction", layout="wide")

# Title
st.title("📊 Store Sales Prediction Dashboard")
st.write("Predict store sales based on store conditions.")

# Layout
col1, col2 = st.columns(2)

# Column 1 (Store Info)
with col1:
    st.subheader("Store Information")

    # Day of Week
    days_map = {
        "Monday": 1,
        "Tuesday": 2,
        "Wednesday": 3,
        "Thursday": 4,
        "Friday": 5,
        "Saturday": 6,
        "Sunday": 7
    }

    selected_day = st.selectbox("Day Of Week", list(days_map.keys()))
    DayOfWeek = days_map[selected_day]

    # Customers
    Customers = st.number_input(
        "Number of Customers",
        min_value=0,
        max_value=50000,
        value=500
    )

    # School Holiday
    SchoolHoliday_dict = {
        "True": 1,
        "False": 0
    }

    selected_SchoolHoliday = st.selectbox(
        "School Holiday",
        list(SchoolHoliday_dict.keys())
    )
    SchoolHoliday = SchoolHoliday_dict[selected_SchoolHoliday]

    # Year
    Year = st.selectbox("Year", [2013, 2014, 2015])


# Column 2 (Promo Info)
with col2:
    st.subheader("Promotion Information")

    # Month
    Month = st.selectbox("Month", list(range(1, 13)))

    # Promo
    is_promo_dict = {
        "True": 1,
        "False": 0
    }

    selected_is_promo = st.selectbox(
        "Is Promo",
        list(is_promo_dict.keys())
    )
    IsPromo = is_promo_dict[selected_is_promo]

    # Promo2
    selected_is_promo2 = st.selectbox(
        "Is Promo2",
        list(is_promo_dict.keys())
    )
    IsPromo2 = is_promo_dict[selected_is_promo2]

    # Competition Duration
    CompetitionDurationMonths = st.number_input(
        "Competition Duration in Months",
        min_value=0.0,
        value=12.0
    )
    CompetitionDurationMonths_log = np.log1p(CompetitionDurationMonths)


st.write("")


# Prediction
if st.button("Predict Sales"):

    input_data = pd.DataFrame({
        "DayOfWeek": [DayOfWeek],
        "Customers": [np.log1p(Customers)],
        "SchoolHoliday": [SchoolHoliday],
        "Year": [Year],
        "Month": [Month],
        "IsPromo": [IsPromo],
        "IsPromo2": [IsPromo2],
        "CompetitionDurationMonths_log": [CompetitionDurationMonths_log]
    })

    st.write("Input Data:", input_data)

    # Prediction
    prediction = model.predict(input_data)
    prediction_value = np.expm1(prediction)[0]


    # Show Result
    st.markdown(f"""
    ### 💰 Predicted Sales
    # ${prediction_value:,.2f}
    """)

    # calculate avg of customer
    if Customers > 0:
        avg_sale = prediction_value / Customers

        st.info(f"🧍 Average Sale per Customer: ${avg_sale:,.2f}")

        # Average Assessment
        if avg_sale < 5:
            st.warning("⚠️ Low sales per customer")
        elif avg_sale <= 20:
            st.success("✅ Normal sales range")
        else:
            st.error("🚨 Unusually high sales per customer")

    #  Insights to increase sales
    st.subheader("📊 Insights")

    if IsPromo == 1:
        st.write("📈 Promo is active → expected increase in sales")
    else:
        st.write("📉 No promo → sales might be lower")
        st.write("💡 Tip: Activating a promo could increase sales")

    if Customers < 200:
        st.write("👥 Low customer traffic → lower expected sales")
    elif Customers > 1000:
        st.write("🔥 High customer traffic → strong sales expected")

    if CompetitionDurationMonths < 6:
        st.write("🏪 New competition nearby → possible negative impact")
    else:
        st.write("🏪 Competition stabilized → less impact on sales")
```
